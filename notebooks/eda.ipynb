```python
import sys; sys.path.append("..")
import numpy as np, matplotlib.pyplot as plt
from data import get_splits
d = get_splits()
X, y = d["X_train"], d["y_train"]
```

```python
fig, ax = plt.subplots(3, 10, figsize=(12, 4))
for c in range(10):
    for r, i in enumerate(np.where(y == c)[0][:3]):
        ax[r, c].imshow(X[i].reshape(28, 28), cmap="gray"); ax[r, c].axis("off")
plt.savefig("../report/eda_samples.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
plt.figure()
plt.bar(range(10), np.bincount(y)); plt.xticks(range(10))
plt.xlabel("Chữ số"); plt.ylabel("Số mẫu")
plt.savefig("../report/eda_classes.png", dpi=150)
plt.show()
```

```python
fig, ax = plt.subplots(1, 11, figsize=(14, 2))
for c in range(10):
    ax[c].imshow(X[y == c].mean(0).reshape(28, 28), cmap="gray"); ax[c].axis("off"); ax[c].set_title(c)
ax[10].imshow(X.var(0).reshape(28, 28), cmap="hot"); ax[10].axis("off"); ax[10].set_title("var")
plt.savefig("../report/eda_mean_var.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
print("Tỉ lệ pixel bằng 0:", (X == 0).mean())
print("Số pixel luôn bằng 0:", (X.var(0) == 0).sum())
print("Khoảng giá trị:", X.min(), X.max())
```
